```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# loading and readign the dataset
Train_data = np.loadtxt('train.csv', delimiter=',')
Test_data = np.loadtxt('test.csv', delimiter=',')
```

```python
Train_data.shape, Test_data.shape
```

```python
train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')
```

```python
train_data.head(), test_data.head()
```

```python
X = Train_data[:, :10]
Y = Train_data[:, 10:]
```

```python
print(X.shape, Y.shape)
```

```python
# Splitting the dataset into training and testing sets
# 15% of the data will be used for testing
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.15)
```

```python
# score function based on coefficient of determination
def score(y_true, y_pred):
    return 1 - np.sum((y_true - y_pred) **2) / np.sum((y_true - np.mean(y_true)) ** 2)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Coefficient of Determination

```python
alg1 = LinearRegression()
alg1.fit(x_train, y_train)
y_pred = alg1.predict(x_test)
print("-------Testing Score-------")
print(score(y_test,y_pred))
```

```python
print("-------Training Score-------")
y_pred_train = alg1.predict(x_train)
print(score(y_train, y_pred_train))
```

```python
X_Test = Test_data
X_train = Train_data[:, :10]
Y_train = Train_data[:, 10:]
```
