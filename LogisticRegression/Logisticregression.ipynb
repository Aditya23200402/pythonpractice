```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
train.drop(['Ticket', 'Cabin'], axis=1, inplace=True)
test.drop(['Ticket', 'Cabin'], axis=1, inplace=True)

# Handling missing values and encoding categorical variables

for df in [train, test]:
    df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
    df['Age'].fillna(df['Age'].median(), inplace=True)
    df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)


for df in [train, test]:
    df['Embarked'] = df['Embarked'].astype(str).str.strip()
    df['Embarked'].replace('', np.nan, inplace=True)


train['Sex'].fillna(train['Sex'].mode()[0], inplace=True)
test['Sex'].fillna(test['Sex'].mode()[0], inplace=True)

map_sex = {'male': 0, 'female': 1}
map_embarked = {'S': 0, 'C': 1, 'Q': 2}

# train['Sex'] = train['Sex'].map(map_sex)
# train['Embarked'] = train['Embarked'].map(map_embarked)

for df in [train, test]:
    df['Sex'] = df['Sex'].map(map_sex)
    df['Embarked'] = df['Embarked'].map(map_embarked)

x_train = train[features]
y_train = train['Survived']
x_test = test[features]

print(x_train.isnull().sum())

# scaler = StandardScaler()
# x_train = scaler.fit_transform(x_train)
# x_test = scaler.transform(x_test)

# model = LogisticRegression()
# model.fit(x_train, y_train)
# predictions = model.predict(x_test)
```
