Links of all the Pages

```python
all_urls = ['https://books.toscrape.com/catalogue/page-1.html']
current_page = 'https://books.toscrape.com/catalogue/page-1.html'

base_url = 'https://books.toscrape.com/catalogue/'

response = requests.get(current_page)

while response.status_code == 200:
    data = BeautifulSoup(response.text, 'html.parser')
    next_page = data.find(class_ = 'next')
    if next_page is None:
        break
    next_page_url = base_url + next_page.a['href']
    print(next_page_url)
    all_urls.append(next_page_url)
    current_page = next_page_url
    response = requests.get(current_page)
```

Find the Image with maximum area

```python
res = requests.get('https://en.wikipedia.org/wiki/Artificial_intelligence')
data = BeautifulSoup(res.text,"html.parser")
Image = data.find_all('img')
max_area = 0
max_image = ""
for img in Image:
    if img.has_attr('width') and img.has_attr('height'):
        area = float(img['width']) * float(img['height'])
        if area > max_area:
            max_area = area
            max_image = img['src']
print(max_image)
```

Quotes with tag Humour

```python
from bs4 import BeautifulSoup
import requests
url='http://quotes.toscrape.com/'
BaseUrl ='http://quotes.toscrape.com/'
quotes = []
res = requests.get(url)
while res.status_code == 200:
    data = BeautifulSoup(res.text,'html.parser')
    for i in data.find_all(class_='quote'):

        for j in i.find_all(class_='tag'):

            if j.text == 'humor':
                quotes.append(i.find(class_='text').text)
    next_page = data.find(class_='next')
    if next_page is None:
        break
    next_page_url = BaseUrl + next_page.a['href']
    res = requests.get(next_page_url)
for i in quotes:
    print(i)
```

Print all authors

```python
from bs4 import BeautifulSoup
import requests
url='http://quotes.toscrape.com/'
BaseUrl ='http://quotes.toscrape.com/'
authors = []
res = requests.get(url)
while res.status_code == 200:
   data = BeautifulSoup(res.text, 'html.parser')
   for i in data.find_all(class_='author'):
       authors.append(i.text)
   if data.find(class_='next') is None:
       break
   next_page = data.find(class_='next').find('a')['href']
   url = BaseUrl + next_page
   res = requests.get(url)

temp = []
for i in sorted(authors):
    if i not in temp:
        print(i)
    temp.append(i)
```

Birth date of authors

```python
import requests
from bs4 import BeautifulSoup

authors = {}
for i in range(1,11):
    response = requests.get('http://quotes.toscrape.com/page/' + str(i) + '/')
    data = BeautifulSoup(response.text, 'html.parser')
    for aut in data.select('.author') :
        if aut.text[0] == 'J':
            authors[aut.text] = aut.next_sibling.next_sibling['href']

bdate = {}
for author in sorted(authors):
    page = requests.get('http://quotes.toscrape.com' + authors[author])
    data = BeautifulSoup(page.text, 'html.parser')
    for i in data.select('.author-born-date'):
        bdate[author] = i.text
print(bdate)

```

Quotes by Albert Einstein

```python
from bs4 import BeautifulSoup
import requests
url='http://quotes.toscrape.com/'
BaseUrl ='http://quotes.toscrape.com/'
authors = []
res = requests.get(url)
while res.status_code == 200:
   data = BeautifulSoup(res.text, 'html.parser')
   for i in data.find_all(class_='quote'):
      s = i.find(class_='author').string
      if s == 'Albert Einstein' :
         authors.append(i.find(class_='text').string)
   if data.find(class_='next') is None:
       break
   next_page = data.find(class_='next').find('a')['href']
   url = BaseUrl + next_page
   res = requests.get(url)
for i in authors:
   print(i)
```
