```python
import numpy as np
import pandas as pd
from sklearn import model_selection
from sklearn import preprocessing
training = np.loadtxt('train.csv', delimiter=',')
testing = np.loadtxt('test.csv', delimiter=',')
```

```python
def step_gradient(m, x, y, leanrning_rate):
    m_slope = np.zeros(len(x[0])) # Initialize slope
    for i in range(len(x)):
        X = x[i]
        Y = y[i]
        for j in range(len(X)):
            m_slope[j] += (-2/len(x) * (Y - sum(m * X)) * X[j])
        new_m = m - (leanrning_rate * m_slope)
    return new_m

def cost(m,x,y):
    cost=0
    for i in range(len(x)):
        cost += (1/len(x))*((y[i] - sum(m * x[i]))**2)
    print(cost)

def gd(x, y, learning_rate, iterations):
    m = np.zeros(len(x[0])) # Initialize weights
    for i in range(iterations):
        m = step_gradient(m, x, y, learning_rate)
        print("itr=", i, "cost=", end=' ')
        cost(m,x,y)
    return m


def gradient_descent(x,y):
    learning_rate = 0.005
    iterations = 500
    x=np.append(x, np.ones(len(x)).reshape(-1, 1), axis=1) # Add bias term
    m = gd(x,y, learning_rate, iterations)
    return m
```

```python
x_train = training[:, :-1] # features
y_train = training[:, -1] # target variable

# scaling
scaler = preprocessing.StandardScaler()
scaler.fit(x_train)
x_train = scaler.transform(x_train)
m = gradient_descent(x_train, y_train)
```

```python
# testing the model on test data

x_test = scaler.transform(testing)
x_test = np.append(x_test, np.ones(len(x_test)).reshape(-1,1), axis=1)
def predict(x, m):
    y_pred = []
    for i in range(len(x)):
        y_pred.append(np.dot(m, x[i]))
    return np.array(y_pred)

y_pred = predict(x_test, m)
print(y_pred)
```
